### Explicar la diferencia entre len(texto), una función, y texto.lower(), un método.
Función vs Método

len(texto) es una función. Existe por sí sola, no pertenece a ningún tipo de dato, y el texto se le pasa como argumento entre paréntesis. Sirve para muchos tipos: cadenas, listas, etc.

texto.lower() es un método. Pertenece al tipo str y se llama desde el objeto con un punto. Solo existe para cadenas: [1, 2].lower() da error porque una lista no tiene ese método.

```python
# ANTES: línea original, tal como viene en registros
linea = registros[0]
print("ANTES:  ", repr(linea))

# LIMPIEZA
campos = linea.split("|")                      # separar por |
campos = [campo.strip() for campo in campos]   # quitar espacios de los extremos
campos[0] = campos[0].upper()                  # identificador en mayúsculas
campos[2] = campos[2].lower()                  # correo en minúsculas
campos[3] = campos[3].lower()                  # categoría en minúsculas

# DESPUÉS: registro transformado
print("DESPUÉS:", campos)

# El original no se modificó (se conserva para auditoría)
print("ORIGINAL:", repr(registros[0]))
```

**Transformación que no se aplicaría automáticamente a nombres propios:** `capitalize()` (o `lower()`).

Razón: altera la capitalización original de nombres y siglas (por ejemplo, "McDonald" o "IA"), y esa información no se puede recuperar después. Por eso el nombre se conserva tal cual y solo se normalizan el identificador, el correo y la categoría.

## Actividad 4
### Funciones reutilizables

```python
def limpiar_texto(texto):
    return " ".join(texto.strip().split())

def correo_basico(correo):
    if correo.count("@") != 1 or " " in correo:
        return False
    usuario, dominio = correo.split("@")
    partes = dominio.split(".")
    return (usuario != "" and len(partes) >= 2
            and all(parte != "" for parte in partes))

categorias = ["datos", "programacion", "visualizacion"]

def procesar_registro(linea):
    campos = linea.split("|")
    if len(campos) != 5:
        return None, "numero de campos"
    campos = [limpiar_texto(c) for c in campos]
    ticket, nombre, correo, categoria, tiempo = campos
    ticket = ticket.upper()
    correo = correo.lower()
    categoria = categoria.lower()
    if not (len(ticket) == 4 and ticket.startswith("T")
            and ticket[1:].isascii()
            and ticket[1:].isdigit()):
        return None, "identificador"
    if nombre == "":
        return None, "nombre vacio"
    if not correo_basico(correo):
        return None, "correo"
    if categoria not in categorias:
        return None, "categoria"
    if not (tiempo.isascii() and tiempo.isdigit()):
        return None, "minutos no enteros o negativos"
    minutos = int(tiempo)
    if minutos > 180:
        return None, "minutos fuera de rango"
    return [ticket, nombre, correo, categoria, minutos], ""
```

```python
aceptados = []
rechazados = []
for linea in registros:
    fila, motivo = procesar_registro(linea)
    if fila is None:
        rechazados.append([linea, motivo])
    else:
        aceptados.append(fila)

print("Aceptados:", len(aceptados))
print("Rechazados:", len(rechazados))
for linea, motivo in rechazados:
    print(repr(linea), "->", motivo)
```

```python
# Pruebas
assert len(aceptados) + len(rechazados) == len(registros)
assert len(aceptados) == 8
assert len(rechazados) == 4
assert procesar_registro("T100|Ana|a@b.org|datos|0")[0]
assert procesar_registro("T100|Ana|a@b.org|datos|180")[0]
assert procesar_registro("T100|Ana|a@b.org|datos|181")[0] is None
assert procesar_registro("T100|Ana|ana@@b.org|datos|5")[0] is None
assert procesar_registro("T100|Ana|ana@b..org|datos|5")[0] is None
assert procesar_registro("T100|Ana|a@b.org|datos")[0] is None
assert procesar_registro("T100|Ana|a@b.org|musica|5")[0] is None
print("Todas las pruebas pasaron")

# Si no aparece ningún error, todo está bien.
# Las últimas cuatro pruebas cubren los casos de la tabla de casos límite, la línea de cuatro campos y la categoría desconocida.
```

## Actividad 5
### Indicadores y gráfica

```python
# Total de registros de entrada
print("Total registros de entrada:", len(registros))

# Número de aceptados
print("Número de aceptados:",(len(aceptados)))

# Número de rechazados
print("Número de rechazados:", len(rechazados))

# Porcentaje de aceptación
n_validos = len(aceptados)
n_total = len(registros)
porcentaje = 100 * n_validos / n_total
print(f"Porcentaje de aceptación: {porcentaje:.2f}%")

# Media de minutos entre aceptados
tiempos = [fila[4] for fila in aceptados]
n = len(aceptados)
promedio = sum(tiempos) / n if n else 0
print(f"Tiempo promedio: {promedio:.2f} min")

# Número de solicitudes por categoría
conteos = []
for categoria in categorias:
    cantidad = 0
    for fila in aceptados:
        if fila[3] == categoria:
            cantidad += 1
    conteos.append(cantidad)
print("Número de solicitudes por categoría:", conteos)

# Clasificación conteos
en_tiempo = 0
fuera_de_tiempo = 0
limite = 15
for fila in aceptados:
    if fila[4] <= limite:
        en_tiempo += 1
    else:
        fuera_de_tiempo += 1
print("En tiempo:", en_tiempo)
print("Fuera de tiempo:", fuera_de_tiempo)
```

```python
def graficar_barras(etiquetas, valores, titulo, color):
    fig, ax = plt.subplots(figsize=(6, 3.5))
    ax.bar(etiquetas, valores, color=color)
    ax.set_title(titulo)
    ax.set_xlabel("Categoria")
    ax.set_ylabel("Registros aceptados")
    fig.tight_layout()
    return fig, ax

fig, ax = graficar_barras(categorias, conteos,
                          "Solicitudes por categoria", "steelblue")
fig.savefig("solicitudes_por_categoria.png", dpi=150)
plt.show()
```
