### Diferencia entre (len(registros)) y (len(registros[0]))
En (len(registros)) cuenta la longitud de elementos en la lista, mientras que (len(registros[0])) cuenta la longitud de caracteres del índice 0 de la lista.

```python
print(registros[0])
print(registros[:3])
print(registros[-1])
print()
print(len(registros))
print(len(registros[0]))
```

### "T001" in registros y "T001" in registros[0]
En "T001" in registros, Python recorre la lista y compara "T001" contra cada elemento completo. Python devolverá False, porque ninguna línea completa es igual a "T001". En cuanto a "T001" in registros[0], Python buscará dentro de la cadena registros[0] si "T001" aparece dentro del texto. Al sí aparecer dentro, Python devolverá True.

```python
print("T001" in registros)
print("T001" in registros[0])
```
